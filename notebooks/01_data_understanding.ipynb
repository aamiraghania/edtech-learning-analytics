#### **Import library dan dataset**

```python
import pandas as pd
from pathlib import Path

base_path = Path(
    "/content/drive/MyDrive/Data_Analytics/"
    "edtech-learning-analytics/data/raw"
)

student_info = pd.read_csv(base_path / "studentInfo.csv")
student_vle = pd.read_csv(base_path / "studentVle.csv")
student_registration = pd.read_csv(base_path / "studentRegistration.csv")
courses = pd.read_csv(base_path / "courses.csv")

print("Dataset berhasil dimuat")
```

#### **Memahami struktur data**

```python
datasets = {
    "student_info": student_info,
    "student_vle": student_vle,
    "student_registration": student_registration,
    "courses": courses
}

for name, df in datasets.items():
  print(f"\n===== {name} =====")
  print("Jumlah baris:", df.shape[0])
  print("Jumlah kolom:", df.shape[1])
  print("\nTipe data:")
  print(df.dtypes)
  print("\nMissing values:")
  print(df.isnull().sum())
```

## **Pemahaman Awal Data**

**1. Gambaran Umum Dataset**

Dataset yang digunakan adalah Open University Learning Analytics Dataset (OULAD), yang memuat informasi mahasiswa, aktivitas pembelajaran digital, pendaftaran, dan hasil akhir pembelajaran.

Empat tabel yang digunakan:
- **studentInfo:** Informasi mahasiswa dan hasil pembelajaran (32.593 baris, 12 kolom).
- **studentVle:** Aktivitas interaksi mahasiswa dengan platform pembelajaran (10.655.280 baris, 6 kolom).
- **studentRegistration:** Informasi pendaftaran dan pengunduran diri (32.593 baris, 5 kolom).
- **courses:** Informasi modul dan periode pembelajaran (22 baris, 3 kolom).

**2. Penilaian Kualitas Data**

Berdasarkan pemeriksaan awal terhadap keempat dataset, diperoleh hasil berikut:

- **Jumlah baris dan kolom:** Keempat dataset memiliki ukuran yang berbeda, dengan studentVle sebagai dataset terbesar, yaitu 10.655.280 baris dan 6 kolom.
- **Missing values:** Tidak ditemukan missing values berdasarkan pemeriksaan menggunakan isnull().sum(). Namun, perlu dilakukan pemeriksaan lebih lanjut terhadap kemungkinan nilai kosong yang direpresentasikan sebagai teks atau simbol tertentu.
- **Tipe data:** Sebagian besar kolom memiliki tipe data yang sesuai, yaitu object untuk data kategorikal dan int64 untuk data numerik. Kolom date_registration dan date_unregistration masih bertipe object sehingga perlu diperiksa lebih lanjut.

**3. Temuan Awal**

Berdasarkan distribusi final_result, diperoleh hasil berikut:

- **Pass:** 37,93%
- **Distinction:** 9,28%
- **Fail:** 21,64%
- **Withdrawn:** 31,16%

**Interpretasi awal:**

Sebagian besar observasi mahasiswa-modul memperoleh hasil *Pass* (37,93%), diikuti oleh *Withdrawn* (31,16%), *Fail* (21,64%), dan *Distinction* (9,28%). Persentase *Withdrawn* yang mencapai 31,16% menunjukkan bahwa pengunduran diri merupakan salah satu aspek penting untuk dianalisis lebih lanjut.
